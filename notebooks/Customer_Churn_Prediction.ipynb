# Logistic Regression from scratch

```python
import numpy as np

class LogisticRegressionScratch:
 
    def __init__(self, learning_rate=0.01, n_iterations=1000):
        self.learning_rate = learning_rate
        self.n_iterations  = n_iterations
        self.weights = None
        self.bias    = None
        self.train_losses = []
        self.val_losses   = []
    
    def _sigmoid(self, z):
        z = np.clip(z, -500, 500)
        return 1 / (1 + np.exp(-z))
    
    def _loss(self, y_true, y_pred):
        epsilon = 1e-15
        y_pred = np.clip(y_pred, epsilon, 1 - epsilon)
        m = len(y_true)
        return -(1/m) * np.sum(
            y_true * np.log(y_pred) + (1 - y_true) * np.log(1 - y_pred)
        )
    
    def fit(self, X_train, y_train, X_val=None, y_val=None):
        X_train = np.array(X_train)
        y_train = np.array(y_train)
        if X_val is not None:
            X_val = np.array(X_val)
            y_val = np.array(y_val)
        
        n_features = X_train.shape[1]
        self.weights = np.zeros(n_features)
        self.bias = 0
        m = X_train.shape[0]
        
        self.history = []

        for i in range(self.n_iterations):
        
            z_train = np.dot(X_train, self.weights) + self.bias
            y_pred_train = self._sigmoid(z_train)
            
            # Compute losses
            train_loss = self._loss(y_train, y_pred_train)
            self.train_losses.append(train_loss)
            
            if X_val is not None:
                z_val = np.dot(X_val, self.weights) + self.bias
                y_pred_val = self._sigmoid(z_val)
                val_loss = self._loss(y_val, y_pred_val)
                self.val_losses.append(val_loss)
            
            error = y_pred_train - y_train
            dw = (1/m) * np.dot(X_train.T, error)
            db = (1/m) * np.sum(error)
            
            self.weights -= self.learning_rate * dw
            self.bias    -= self.learning_rate * db
            
          
            if (i + 1) % 100 == 0:
                if X_val is not None:
                    print(f"Iteration {i+1:4d} | Train Loss: {train_loss:.4f} | Val Loss: {val_loss:.4f}")
                    
                    self.history.append({
                         "Iteration": i + 1,
                         "Train Loss": round(train_loss, 4),
                         "Val Loss": round(val_loss, 4)
                    })
                else:
                    print(f"Iteration {i+1:4d} | Train Loss: {train_loss:.4f}")
                    
                    self.history.append({
                         "Iteration": i + 1,
                         "Train Loss": round(train_loss, 4),
                  
                    })
                    
        return self
    
    def predict_proba(self, X):
        """Predict probabilities."""
        X = np.array(X)
        z = np.dot(X, self.weights) + self.bias
        return self._sigmoid(z)
    
    def predict(self, X, threshold=0.5):
        """Predict binary labels (0 or 1)."""
        probabilities = self.predict_proba(X)
        return (probabilities >= threshold).astype(int)
```

## Train the Model

```python
# initialize the model 
model_scratch = LogisticRegressionScratch(
    learning_rate=0.01,
    n_iterations=1500
)

# Train on training data, track validation loss
model_scratch.fit(X_train, y_train, X_val, y_val)

print("\n complete")
pd.DataFrame(model_scratch.history)
```

```python
# Check parameters
params_table = pd.DataFrame({
    "Parameter": [
        "Number of iterations",
        "Learning rate",
        "Number of features",
        "Final bias",
        "Final Train Loss",
        "Final Val Loss"
    ],
    "Value": [
        model_scratch.n_iterations,
        model_scratch.learning_rate,
        len(model_scratch.weights),
        round(model_scratch.bias, 4),
        round(model_scratch.train_losses[-1], 4),
        round(model_scratch.val_losses[-1], 4)
    ]
})

print("\nModel Parameters Table:")
params_table
```

```python
# Map weights to feature names
feature_importance = pd.DataFrame({
    'Feature': X_train.columns,
    'Weight':  model_scratch.weights
}).sort_values('Weight', key=abs, ascending=False)

print("Top 10 most influential features:")
feature_importance.head(10)
```

# Logistic Regression (sklearn)

```python
from sklearn.linear_model import LogisticRegression

model_sklearn = LogisticRegression(
max_iter=1500,
random_state=42
)

# Train on training data
model_sklearn.fit(X_train, y_train)

print("sklearn Logistic Regression trained successfully")
print(f"\nNumber of iterations used: {model_sklearn.n_iter_[0]}")
print(f"Final intercept (bias):   {model_sklearn.intercept_[0]:.4f}")
```

```python
# Compare weights between scratch and sklearn implementations

comparison_df = pd.DataFrame({
'Feature':        X_train.columns,
'Scratch Weight': model_scratch.weights,
'sklearn Weight': model_sklearn.coef_[0]
})


comparison_df['|Difference|'] = abs(
comparison_df['Scratch Weight'] - comparison_df['sklearn Weight']
)

comparison_df = comparison_df.reindex(
comparison_df['sklearn Weight'].abs().sort_values(ascending=False).index
)

print("Top 10 features — weight comparison:")
comparison_df.head(10).round(4)
```

```python
from sklearn.metrics import log_loss

# Get predictions 
y_pred_train_sk = model_sklearn.predict_proba(X_train)[:, 1]
y_pred_val_sk   = model_sklearn.predict_proba(X_val)[:, 1]

# Compute losses
train_loss_sk = log_loss(y_train, y_pred_train_sk)
val_loss_sk   = log_loss(y_val,   y_pred_val_sk)

# Comparison 
comparison = pd.DataFrame({
    'Metric': ['Bias ', 'Train Loss', 'Validation Loss'],
    'Scratch': [
        round(model_scratch.bias, 4),
        round(model_scratch.train_losses[-1], 4),
        round(model_scratch.val_losses[-1], 4)
    ],
    'sklearn': [
        round(model_sklearn.intercept_[0], 4),
        round(train_loss_sk, 4),
        round(val_loss_sk, 4)
    ]
})

print("Scratch vs sklearn comparison:")
comparison
```

# Unified Model Evaluation

```python
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, 
    f1_score, confusion_matrix, classification_report
)

def evaluate_model(name, y_true, y_pred):
   
    return {
        'Model':     name,
        'Accuracy':  accuracy_score(y_true, y_pred),
        'Precision': precision_score(y_true, y_pred),
        'Recall':    recall_score(y_true, y_pred),
        'F1-Score':  f1_score(y_true, y_pred)
    }
```

```python
y_pred_scratch  = model_scratch.predict(X_val)
y_pred_sklearn  = model_sklearn.predict(X_val)
y_pred_rf       = model_rf.predict(X_val)
```

```python
# Evaluate all three models
results = [
    evaluate_model('LogReg (Scratch)', y_val, y_pred_scratch),
    evaluate_model('LogReg (sklearn)', y_val, y_pred_sklearn),
    evaluate_model('Random Forest',    y_val, y_pred_rf)
]


results_df = pd.DataFrame(results)

print("Model Comparison on Validation Set:")
print("="*60)
results_df.round(4)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Compute confusion matrices
cm_scratch  = confusion_matrix(y_val, y_pred_scratch)
cm_sklearn  = confusion_matrix(y_val, y_pred_sklearn)
cm_rf       = confusion_matrix(y_val, y_pred_rf)


fig, axes = plt.subplots(1, 3, figsize=(16, 5))

models_cms = [
   ('LogReg (Scratch)', cm_scratch, axes[0]),
   ('LogReg (sklearn)', cm_sklearn, axes[1]),
   ('Random Forest',    cm_rf,      axes[2])
]

for name, cm, ax in models_cms:
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
                xticklabels=['No Churn', 'Churn'],
                yticklabels=['No Churn', 'Churn'],
                ax=ax, cbar=False, annot_kws={'size': 14})
    ax.set_title(f'{name}', fontsize=13, fontweight='bold')
    ax.set_xlabel('Predicted', fontsize=11)
    ax.set_ylabel('Actual', fontsize=11)

plt.suptitle('Confusion Matrices Comparison (Validation Set)',
             fontsize=14, fontweight='bold', y=1.02)
plt.tight_layout()
plt.savefig('confusion_matrices.png', dpi=300, bbox_inches='tight')
plt.show()
```

# Hyperparameter Tuning (Random Forest)

```python
from sklearn.model_selection import GridSearchCV

param_grid = {
    'max_depth':         [5, 10, 15, 20],
    'min_samples_split': [2, 5, 10],
    'n_estimators':      [50, 100, 200]
}


rf_base = RandomForestClassifier(random_state=42, n_jobs=-1)


grid_search = GridSearchCV(
    estimator=rf_base,
    param_grid=param_grid,
    cv=5,                   
    scoring='f1',          
    n_jobs=-1,            
    verbose=1 
)

print("Starting Grid Search")
grid_search.fit(X_train, y_train)



print(f" Best Parameters: {grid_search.best_params_}")
print(f" Best CV F1-Score: {grid_search.best_score_:.4f}")
```

```python
from sklearn.metrics import log_loss


model_rf_tuned = grid_search.best_estimator_


y_proba_train_tuned = model_rf_tuned.predict_proba(X_train)[:, 1]
y_proba_val_tuned   = model_rf_tuned.predict_proba(X_val)[:, 1]
train_loss_tuned    = log_loss(y_train, y_proba_train_tuned)
val_loss_tuned      = log_loss(y_val,   y_proba_val_tuned)

# Comparison Before vs After Tuning
comparison_tuning = pd.DataFrame({
    'Metric': ['Train Loss', 'Validation Loss', 'Gap (Overfitting)'],
    'RF (Default)': [
        round(train_loss_rf, 4),
        round(val_loss_rf, 4),
        round(abs(val_loss_rf - train_loss_rf), 4)
    ],
    'RF (Tuned)': [
        round(train_loss_tuned, 4),
        round(val_loss_tuned, 4),
        round(abs(val_loss_tuned - train_loss_tuned), 4)
    ]
})

print("Random Forest: Before vs After Tuning")
print("="*55)
comparison_tuning
```
